```python
groups = []

for token_id, token, (start, end) in zip(input_ids, tokens, offsets):
    if groups and start < groups[-1]["end"]:
        groups[-1]["end"] = max(groups[-1]["end"], end)
        groups[-1]["ids"].append(token_id)
        groups[-1]["tokens"].append(token)
    else:
        groups.append({
            "start": start,
            "end": end,
            "ids": [token_id],
            "tokens": [token],
        })

for i, g in enumerate(groups):
    piece = chinese_text[g["start"]:g["end"]]
    print(f"{i:02d} | text={piece!r} | ids={g['ids']} | tokens={g['tokens']} | span=({g['start']}, {g['end']})")
```
